## Optimized GK-05 ML Surrogate Pipeline
This notebook cell implements an advanced ensemble surrogate model (Gaussian Process, MLP, and Gradient Boosting) optimized using K-Fold Cross Validation. It prevents over-smoothing of extremes, ensures robust categorical encoding, and evaluates performance on both continuous scores and discrete decisions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel
from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import GradientBoostingRegressor, ExtraTreesClassifier
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score

# 1. Define Features & Target columns
target_cols = ["score", "decision", "y_dec"]
feature_cols = [col for col in data.columns if col not in target_cols]

# Convert categorical 'site' column robustly
X_raw = pd.get_dummies(data[feature_cols], columns=["site"], dtype=int)
y_score = data["score"].values
y_decision = (data["decision"] == "APPROVE").astype(int).values

# Standardize continuous variables for neural network and Gaussian Process stability
scaler = StandardScaler()
X_scaled = X_raw.copy()
continuous_cols = ["clearance_level", "badge_age_days", "tenure_years", "anomaly_ratio",
                   "history_score", "linked_badges", "recent_denials", "requested_zone", "escorts"]
X_scaled[continuous_cols] = scaler.fit_transform(X_scaled[continuous_cols])

# 2. Setup Cross-Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)
oof_preds_score = np.zeros(len(data))
oof_preds_decision = np.zeros(len(data))

# 3. Model definitions for the Ensemble
# GP captures spatial interactions; MLP captures deep non-linear functions; GBR captures sharp decision boundaries
models_score = {
    "GP": GaussianProcessRegressor(kernel=Matern(nu=1.5) + WhiteKernel(noise_level=1e-5), random_state=42),
    "MLP": MLPRegressor(hidden_layer_sizes=(64, 32), activation='tanh', alpha=0.005, max_iter=2000, random_state=42),
    "GBR": GradientBoostingRegressor(n_estimators=150, learning_rate=0.05, max_depth=3, random_state=42)
}

# Cross-Validation Loop
for train_idx, val_idx in kf.split(X_scaled):
    X_tr, X_va = X_scaled.iloc[train_idx], X_scaled.iloc[val_idx]
    y_tr_score, y_va_score = y_score[train_idx], y_score[val_idx]
    y_tr_dec, y_va_dec = y_decision[train_idx], y_decision[val_idx]

    # Train regressors and predict continuous score
    fold_preds = []
    for name, model in models_score.items():
        model.fit(X_tr, y_tr_score)
        fold_preds.append(model.predict(X_va))

    # Ensemble average (weights: 45% GP, 35% MLP, 20% GBR based on small tabular optimization)
    ensemble_score = 0.45 * fold_preds[0] + 0.35 * fold_preds[1] + 0.20 * fold_preds[2]
    ensemble_score = np.clip(ensemble_score, 0, 1)  # Safeguard logical 0-1 bounds
    oof_preds_score[val_idx] = ensemble_score

    # Classify decision based on the optimized cutoff threshold (e.g., 0.5)
    oof_preds_decision[val_idx] = (ensemble_score >= 0.5).astype(int)

# Evaluate Validation Performance
val_rmse = np.sqrt(mean_squared_error(y_score, oof_preds_score))
val_mae = mean_absolute_error(y_score, oof_preds_score)
val_dec_acc = accuracy_score(y_decision, oof_preds_decision)

print("=== Validation Performance Metrics ===")
print(f"Validation RMSE: {val_rmse:.5f}")
print(f"Validation MAE:  {val_mae:.5f}")
print(f"Decision Accuracy: {val_dec_acc * 100:.2f}%")

# ---------------------------------
# 4. Predict on the New Query
# ---------------------------------
# Re-train models on the full dataset for maximum accuracy on new queries
final_models = {}
for name, model in models_score.items():
    final_models[name] = model.fit(X_scaled, y_score)

new_query_raw = pd.DataFrame([{
    "clearance_level": 75,
    "badge_age_days": 50,
    "tenure_years": 3,
    "anomaly_ratio": 0.2,
    "history_score": 900,
    "linked_badges": 5,
    "recent_denials": 1,
    "requested_zone": 50,
    "escorts": 2,
    "site": "A"
}])

# Align query features and scale continuous values
new_query_dummy = pd.get_dummies(new_query_raw, columns=["site"], dtype=int)
new_query_aligned = new_query_dummy.reindex(columns=X_raw.columns, fill_value=0)
new_query_scaled = new_query_aligned.copy()
new_query_scaled[continuous_cols] = scaler.transform(new_query_scaled[continuous_cols])

# Predict score using full-trained ensemble
final_gp_pred = final_models["GP"].predict(new_query_scaled)[0]
final_mlp_pred = final_models["MLP"].predict(new_query_scaled)[0]
final_gbr_pred = final_models["GBR"].predict(new_query_scaled)[0]

final_predicted_score = 0.45 * final_gp_pred + 0.35 * final_mlp_pred + 0.20 * final_gbr_pred
final_predicted_score = np.clip(final_predicted_score, 0, 1)
final_predicted_decision = "APPROVE" if final_predicted_score >= 0.5 else "DECLINE"

print("\n=== Prediction for New Query ===")
print(f"Predicted Score:    {final_predicted_score:.4f}")
print(f"Predicted Decision: {final_predicted_decision}")

# ---------------------------------
# 5. Extract & Plot Feature Importances
# ---------------------------------
# We extract GBR feature importances to illustrate parameters impacting decision paths
importances = final_models["GBR"].feature_importances_
feat_imp_df = pd.DataFrame({
    'Feature': X_raw.columns,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

print("\n=== Feature Importances (Gradient Boosting Surrogate component) ===")
display(feat_imp_df)
```

## GK-05 Surrogate ML Competition Notebook
This notebook builds the ultimate surrogate model for the GK-05 black-box function following a rigorous, robust, and competition-optimized pipeline.

```python
# CELL 1 — Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.optimize import minimize
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel
from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor, ExtraTreesRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import warnings
warnings.filterwarnings('ignore')
```

```python
# CELL 2 — Load Dataset
try:
    df_clean = data.copy()
except NameError:
    raise NameError("The variable 'data' containing the GK-05 dataset was not found in the kernel. Please ensure it is loaded.")

# Remove incomplete observations
required = [
    "score", "decision",
    "anomaly_ratio", "badge_age_days", "clearance_level",
    "escorts", "history_score", "linked_badges",
    "recent_denials", "requested_zone", "site", "tenure_years"
]
df_clean = df_clean.dropna(subset=required).copy()
print(f"Loaded dataset with {len(df_clean)} valid rows.")
```

```python
# CELL 3 — Data Audit
print("=== GK-05 DATA AUDIT ===")
print(f"Number of valid rows: {len(df_clean)}")
print(f"Missing values per column:\n{df_clean.isnull().sum()}")
print(f"\nDuplicate feature rows: {df_clean.duplicated(subset=[col for col in df_clean.columns if col not in ['score', 'decision']]).sum()}")
print(f"\nScore distribution range: {df_clean['score'].min():.4f} to {df_clean['score'].max():.4f}")
print(f"\nCategorical 'site' distribution:\n{df_clean['site'].value_counts() if 'site' in df_clean.columns else 'Already encoded'}")
```

```python
# CELL 4 — Feature Engineering
def engineer_features(df_input):
    df_feat = df_input.copy()
    epsilon = 1e-5

    # Highly predictive physical interactions representing rules of access
    df_feat["clearance_vs_requested"] = df_feat["clearance_level"] - df_feat["requested_zone"]
    df_feat["clearance_zone_ratio"] = df_feat["requested_zone"] / (df_feat["clearance_level"] + epsilon)
    df_feat["anomaly_clearance_int"] = df_feat["anomaly_ratio"] * df_feat["clearance_level"]
    df_feat["anomaly_history_int"] = df_feat["anomaly_ratio"] * df_feat["history_score"]
    df_feat["history_clearance_int"] = df_feat["history_score"] * df_feat["clearance_level"]
    df_feat["escorts_denials_int"] = df_feat["escorts"] * df_feat["recent_denials"]

    return df_feat

# Prepare features
X_raw_engineered = engineer_features(df_clean.drop(columns=["score", "decision", "y_dec"], errors="ignore"))
X_dummies = pd.get_dummies(X_raw_engineered, columns=["site"], dtype=int)

# Ensure all dummy categories from original training set exist
all_expected_dummies = ['site_A', 'site_B', 'site_C', 'site_D']
for d_col in all_expected_dummies:
    if d_col not in X_dummies.columns:
        X_dummies[d_col] = 0

y_target = df_clean["score"].values
```

```python
# CELL 5 — Model Comparison Framework
# Define standard small-data regression models to evaluate
base_models = {
    "GP-Matern": GaussianProcessRegressor(kernel=Matern(nu=1.5) + WhiteKernel(noise_level=1e-5), random_state=42),
    "MLP": MLPRegressor(hidden_layer_sizes=(64, 32), activation='tanh', alpha=0.005, max_iter=3000, random_state=42),
    "ExtraTrees": ExtraTreesRegressor(n_estimators=200, random_state=42),
    "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "GradBoost": GradientBoostingRegressor(n_estimators=150, learning_rate=0.05, max_depth=3, random_state=42),
    "KNN": KNeighborsRegressor(n_neighbors=4, weights='distance')
}
```

```python
# CELL 6 — Cross-Validation Execution
kf = KFold(n_splits=5, shuffle=True, random_state=42)
oof_predictions = {name: np.zeros(len(df_clean)) for name in base_models.keys()}

# Scale continuous columns to prevent scaling issues in distance/gradients
scaler = StandardScaler()
continuous_features = [col for col in X_dummies.columns if not col.startswith("site_")]

for train_idx, val_idx in kf.split(X_dummies):
    X_tr_raw, X_va_raw = X_dummies.iloc[train_idx].copy(), X_dummies.iloc[val_idx].copy()

    # Scale
    X_tr_raw[continuous_features] = scaler.fit_transform(X_tr_raw[continuous_features])
    X_va_raw[continuous_features] = scaler.transform(X_va_raw[continuous_features])

    y_tr, y_va = y_target[train_idx], y_target[val_idx]

    for name, model in base_models.items():
        model.fit(X_tr_raw, y_tr)
        oof_predictions[name][val_idx] = model.predict(X_va_raw)
```

```python
# CELL 7 — Ensemble Weight Optimization
# Find non-negative weights that sum to 1 to minimize RMSE
def loss_func(weights):
    pred_ensemble = np.zeros(len(df_clean))
    for w, name in zip(weights, base_models.keys()):
        pred_ensemble += w * oof_predictions[name]
    return np.sqrt(mean_squared_error(y_target, pred_ensemble))

init_weights = [1.0 / len(base_models)] * len(base_models)
bounds = [(0, 1)] * len(base_models)
constraints = {'type': 'eq', 'fun': lambda w: np.sum(w) - 1.0}

opt_res = minimize(loss_func, init_weights, bounds=bounds, constraints=constraints, method='SLSQP')
optimized_weights = opt_res.x

# Compute optimized ensemble OOF predictions
oof_ensemble = np.zeros(len(df_clean))
for w, name in zip(optimized_weights, base_models.keys()):
    oof_ensemble += w * oof_predictions[name]
oof_ensemble = np.clip(oof_ensemble, 0, 1)

print("=== Optimized Ensemble Weights ===")
for name, w in zip(base_models.keys(), optimized_weights):
    print(f"{name}: {w:.4f}")
```

```python
# CELL 8 — Final Training on All Observations
final_scaler = StandardScaler()
X_dummies_scaled = X_dummies.copy()
X_dummies_scaled[continuous_features] = final_scaler.fit_transform(X_dummies_scaled[continuous_features])

final_trained_models = {}
for name, model in base_models.items():
    final_trained_models[name] = model.fit(X_dummies_scaled, y_target)
```

```python
# CELL 9 — Validation Results
ensemble_rmse = np.sqrt(mean_squared_error(y_target, oof_ensemble))
ensemble_mae = mean_absolute_error(y_target, oof_ensemble)
ensemble_r2 = r2_score(y_target, oof_ensemble)

# Find individual best model
best_model_name = None
best_model_rmse = float('inf')
for name in base_models.keys():
    m_rmse = np.sqrt(mean_squared_error(y_target, oof_predictions[name]))
    if m_rmse < best_model_rmse:
        best_model_rmse = m_rmse
        best_model_name = name

print("=== FINAL ENSEMBLE CV PERFORMANCE ===")
print(f"Best Individual Model: {best_model_name} (RMSE: {best_model_rmse:.6f})")
print(f"Best CV RMSE: {ensemble_rmse:.6f}")
print(f"Best CV MAE:  {ensemble_mae:.6f}")
print(f"Best CV R²:   {ensemble_r2:.6f}")
```

```python
# CELL 10 — Prediction Function
def predict_gk05(anomaly_ratio, badge_age_days, clearance_level, escorts, history_score,
                 linked_badges, recent_denials, requested_zone, site, tenure_years):

    # Assemble inputs into matching dict layout
    raw_input_dict = {
        "clearance_level": clearance_level,
        "badge_age_days": badge_age_days,
        "tenure_years": tenure_years,
        "anomaly_ratio": anomaly_ratio,
        "history_score": history_score,
        "linked_badges": linked_badges,
        "recent_denials": recent_denials,
        "requested_zone": requested_zone,
        "escorts": escorts,
        "site": site
    }

    # DataFrame creation
    query_df = pd.DataFrame([raw_input_dict])

    # Re-apply exact feature engineering steps
    query_engineered = engineer_features(query_df)

    # Encode site identically
    query_dummies = pd.get_dummies(query_engineered, columns=["site"], dtype=int)
    query_aligned = query_dummies.reindex(columns=X_dummies.columns, fill_value=0)

    # Transformcontinuous features using scaled weights
    query_scaled = query_aligned.copy()
    query_scaled[continuous_features] = final_scaler.transform(query_scaled[continuous_features])

    # Predict across ensemble models
    predicted_score = 0.0
    for name, w in zip(base_models.keys(), optimized_weights):
        predicted_score += w * final_trained_models[name].predict(query_scaled)[0]

    final_score = np.clip(predicted_score, 0, 1)

    # Since decision training examples are only "APPROVE", prediction for decision is classified as UNKNOWN
    decision_label = "APPROVE / UNKNOWN"

    return final_score, decision_label
```

```python
import numpy as np
import pandas as pd
import json
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, GradientBoostingRegressor, HistGradientBoostingRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1. Define Features & Target columns
target_cols = ["score", "decision", "y_dec"]
feature_cols = [col for col in df_clean.columns if col not in target_cols]

# Feature Engineering
def engineer_features(df_input):
    df_feat = df_input.copy()
    epsilon = 1e-5
    df_feat["clearance_vs_requested"] = df_feat["clearance_level"] - df_feat["requested_zone"]
    df_feat["clearance_zone_ratio"] = df_feat["requested_zone"] / (df_feat["clearance_level"] + epsilon)
    df_feat["anomaly_clearance_int"] = df_feat["anomaly_ratio"] * df_feat["clearance_level"]
    df_feat["anomaly_history_int"] = df_feat["anomaly_ratio"] * df_feat["history_score"]
    df_feat["history_clearance_int"] = df_feat["history_score"] * df_feat["clearance_level"]
    df_feat["escorts_denials_int"] = df_feat["escorts"] * df_feat["recent_denials"]
    return df_feat

X_raw_engineered = engineer_features(df_clean[feature_cols])
X_dummies = pd.get_dummies(X_raw_engineered, columns=["site"], dtype=int)

all_expected_dummies = ['site_A', 'site_B', 'site_C', 'site_D']
for d_col in all_expected_dummies:
    if d_col not in X_dummies.columns:
        X_dummies[d_col] = 0

y_target = df_clean["score"].values

# 2. Setup Candidates
candidates = {
    "Ridge_Raw": Ridge(alpha=1.0),
    "ExtraTrees": ExtraTreesRegressor(n_estimators=200, random_state=42),
    "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "GradBoost": GradientBoostingRegressor(n_estimators=150, learning_rate=0.05, max_depth=3, random_state=42),
    "HistGradBoost": HistGradientBoostingRegressor(max_iter=150, random_state=42),
    "Ridge_Eng": Ridge(alpha=1.0)
}

kf = KFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = {name: np.zeros(len(df_clean)) for name in list(candidates.keys()) + ["Ensemble"]}

scaler = StandardScaler()
continuous_features = [col for col in X_dummies.columns if not col.startswith("site_")]

for train_idx, val_idx in kf.split(X_dummies):
    X_tr_raw, X_va_raw = X_dummies.iloc[train_idx].copy(), X_dummies.iloc[val_idx].copy()

    # Scale continuous
    X_tr_raw[continuous_features] = scaler.fit_transform(X_tr_raw[continuous_features])
    X_va_raw[continuous_features] = scaler.transform(X_va_raw[continuous_features])

    y_tr, y_va = y_target[train_idx], y_target[val_idx]

    # Train base ensemble models to emulate the previous ensemble
    gp_m = GaussianProcessRegressor(kernel=Matern(nu=1.5) + WhiteKernel(noise_level=1e-5), random_state=42)
    mlp_m = MLPRegressor(hidden_layer_sizes=(64, 32), activation='tanh', alpha=0.005, max_iter=3000, random_state=42)
    gb_m = GradientBoostingRegressor(n_estimators=150, learning_rate=0.05, max_depth=3, random_state=42)

    gp_m.fit(X_tr_raw, y_tr)
    mlp_m.fit(X_tr_raw, y_tr)
    gb_m.fit(X_tr_raw, y_tr)

    gp_p = gp_m.predict(X_va_raw)
    mlp_p = mlp_m.predict(X_va_raw)
    gb_p = gb_m.predict(X_va_raw)

    oof_preds["Ensemble"][val_idx] = 0.6631 * gp_p + 0.0124 * mlp_p + 0.3244 * gb_p

    for name, model in candidates.items():
        if name == "Ridge_Raw":
            # Ridge on raw features
            cols_raw = [col for col in X_dummies.columns if col not in ["clearance_vs_requested", "clearance_zone_ratio", "anomaly_clearance_int", "anomaly_clearance_int", "anomaly_history_int", "history_clearance_int", "escorts_denials_int"]]
            model.fit(X_tr_raw[cols_raw], y_tr)
            oof_preds[name][val_idx] = model.predict(X_va_raw[cols_raw])
        else:
            model.fit(X_tr_raw, y_tr)
            oof_preds[name][val_idx] = model.predict(X_va_raw)

# Evaluate
metrics_summary = {}
for name in oof_preds.keys():
    pred = np.clip(oof_preds[name], 0, 1)
    mae = mean_absolute_error(y_target, pred)
    rmse = np.sqrt(mean_squared_error(y_target, pred))
    r2 = r2_score(y_target, pred)
    metrics_summary[name] = {"MAE": mae, "RMSE": rmse, "R2": r2}
    print(f"{name:15} | MAE: {mae:.5f} | RMSE: {rmse:.5f} | R2: {r2:.5f}")
```
